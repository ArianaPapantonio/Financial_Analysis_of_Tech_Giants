## Financial Analysis of Tech Giants (2020–2024)

A data-driven analysis of five major tech companies — Apple (AAPL), Amazon (AMZN), Google (GOOGL), Meta (META), and Microsoft (MSFT) — using financial data pulled from a public API.  
This project explores revenue, net income, growth trends, and business efficiency from 2020 to 2024 using Python and data visualization libraries.


```python
import requests
import pandas as pd

API_KEY = "SYoSfbNQdxget24k3u2WzIHBujbMhDWP"
BASE_URL = "https://financialmodelingprep.com/api/v3"
```

```python
# List of tech companies (stock tickers)
companies = ["AAPL", "GOOGL", "MSFT", "AMZN", "META"]
```

```python
# Empty list to store results
financial_data = []
```

```python
# Loop through each company and get income statement
for symbol in companies:
    url = f"{BASE_URL}/income-statement/{symbol}?limit=5&apikey={API_KEY}"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        for record in data:
            financial_data.append({
                "Company": symbol,
                "Date": record.get("date"),
                "Revenue": record.get("revenue"),
                "Net Income": record.get("netIncome")
            })
    else:
        print(f"Failed to fetch data for {symbol}")
```

```python
# Show complete list: 

financial_data
```

```python

# Create a DataFrame
df_financial_data= pd.DataFrame(financial_data)
```

## Data Cleaning

```python
df_financial_data.info
```

```python
# Convert Revenue and Net Income to billions or millions

df_financial_data['Revenue (B)'] = df_financial_data['Revenue'] / 1_000_000_000
df_financial_data['Net Income (B)'] = df_financial_data['Net Income'] / 1_000_000_000
```

```python
df_financial_data.dtypes
```

```python
## Convert Date column to datetime type

df_financial_data['Date'] = pd.to_datetime(df_financial_data['Date'])
```

```python
df_financial_data.dtypes
```

```python
## Create a separate column for month & year

# Create separate columns
df_financial_data['Year'] = df_financial_data['Date'].dt.year
df_financial_data['Month'] = df_financial_data['Date'].dt.month
df_financial_data['Month Name'] = df_financial_data['Date'].dt.strftime('%B')  # Optional: "January", "February", etc.
```

## Data Analysis

```python
## 1. Growth Rate Calculations: Show how each company’s revenue or net income has changed over time
# Perc Growth is calculated substracting last year Revenue (or Net Income) from this year's values, dividing this by last year value and multiplying this resoult x 100 to convert it to a percentage. 

df_financial_data = df_financial_data.sort_values(by=['Company', 'Year']).reset_index(drop=True)  ## Sort DF before calculating growth (so it works from bottom to top)

df_financial_data['Revenue Growth (%)'] = df_financial_data.groupby('Company')['Revenue'].pct_change() * 100
df_financial_data['Net Income Growth (%)'] = df_financial_data.groupby('Company')['Net Income'].pct_change() * 100
```

```python
df_financial_data.head(6)
```

```python
# About results: 
# The first value in each company's group has no "previous value" to compare to, so .pct_change() returns NaN. 
# For this reason each company will always have one NaN in the first row of growth columns.
# For the negative values: this means the metric decreased compared to last year.
```

```python
## 2. Comparison table over the years: Create a pivot table to compare performance year-over-year across companies

# Group by years and sum revenue in billions
df_yearly = df_financial_data.groupby(['Year', 'Company'])['Revenue (B)'].sum().reset_index()

# Create pivot table with billions
pivot = df_yearly.pivot(index='Year', columns='Company', values='Revenue (B)')
```

```python
## 3. Outlier detection: find years where revenue or income dropped significantly

df_financial_data['Revenue Change'] = df_financial_data.groupby('Company')['Revenue'].diff()  # Using Revenue column (Not in billions) for max precision
outliers = df_financial_data[df_financial_data['Revenue Change'] < -1_000_000_000]  # threshold (the rows where revenue dropped more than 1 billion dollars between years)
df_financial_data['Revenue Change (B)'] = df_financial_data['Revenue Change'] / 1_000_000_000
df_financial_data['Revenue Drop?'] = df_financial_data['Revenue Change'].apply(lambda x: 'Yes' if x < -1 else 'No') # Add a 'flag' column for easier tracking
```

```python
df_financial_data.head()
```

```python
## About results: 

# NaN is expected because there’s no previous year to compare (1st row)
# No revenue drop detected — it’s the first data point in the group (1st row)
# 2nd row: Revenue dropped by $7.75 billion; Growth = -1.98% (a small drop in percentage terms);  Net income increased slightly
```

```python
# 4. Rank Companies by Performance: for a given year, rank by revenue (in billions)

latest_year = df_financial_data['Year'].max() # Most recent year (dates do not match for all companies, so using Year instead)
latest_data = df_financial_data[df_financial_data['Year'] == latest_year] # Filters the data to only include records from the latest year — one row per company for that year.
ranked = latest_data.sort_values(by='Revenue (B)', ascending=False) # Sorts the companies in descending order of revenue (biggest at the top).

# View ranked companies
ranked[['Company', 'Revenue (B)', 'Net Income (B)']]
```

```python
# Add numerical ranking 

ranked['Revenue Rank'] = ranked['Revenue (B)'].rank(ascending=False).astype(int)
ranked['Net Income Rank'] = ranked['Net Income (B)'].rank(ascending=False).astype(int)
```

```python
ranked[['Revenue Rank', 'Company', 'Revenue (B)', 'Net Income (B)']]
```

```python
## 5. Business Insight Summary: calculates and stores the average revenue growth for each company across all available years.

growth_data = [] #empty list to store growth info for each company.



for company in companies:
    company_data = df_financial_data[df_financial_data['Company'] == company]
    rev_change = company_data['Revenue'].pct_change().mean() * 100 # Calculate year-over-year revenue growth for the company + calculate the avg (mean) growth rate across all years available and converts to %
    growth_data.append({
        "Company": company,
        "Avg Revenue Growth (%)": round(rev_change, 2) # Prints the company name and its average revenue growth, formatted to 2 decimal places.
    })

df_growth = pd.DataFrame(growth_data) 
df_growth.to_csv("/mnt/c/Users/arian/Documents/Financial_Analysis_of_Tech_Giants/CSV/avg_revenue_growth_by_company.csv", index=False)
```

## Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df_financial_data['Year'].dtype
```

```python
# 1: Revenue over Time (Line Plot)

plt.figure(figsize=(10, 6))

# Create a line plot of revenue over time for each company
sns.lineplot(data=df_financial_data, x="Year", y="Revenue (B)", hue="Company", marker="o") # Separate line for each company & add a dot marker at each data point
plt.title("Revenue Over Time by Company")
plt.ylabel("Revenue (in Billions)")
plt.xlabel("Year")
plt.grid(True)
plt.tight_layout()
plt.xticks(ticks=sorted(df_financial_data['Year'].unique())) #to avoid years showing as float (type is already int)
plt.show()
```

```python
# About the results: 

# Amazon (AMZN) dominates in total revenue, showing consistent growth year over year
# Apple (AAPL) had a strong climb early on, then stabilized, with a slight dip in 2023 and recovery in 2024
# Google (GOOGL) shows steady upward growth — more moderate but consistent
# Microsoft (MSFT) and Meta (META) both grew steadily, with META’s growth accelerating in the last 2 years
```

```python
# 2. Average Revenue Growth (Bar Plot)

# Order DF by average growth in descending order
df_growth = df_growth.sort_values(by="Avg Revenue Growth (%)", ascending=False)

# Adjust colors for each company: 
custom_palette = {
    'AAPL': '#1f77b4',   # blue
    'AMZN': '#ff7f0e',   # orange
    'GOOGL': '#2ca02c',  # green
    'META': '#d62728',   # red
    'MSFT': '#9467bd'    # purple
}
# Create the graph
plt.figure(figsize=(8, 5))
sns.barplot(data=df_growth, x="Avg Revenue Growth (%)", y="Company", palette=custom_palette)

# Add value labels
for index, row in df_growth.iterrows():
    plt.text(
        x=row["Avg Revenue Growth (%)"] + 0.2,
        y=index,
        s=f'{row["Avg Revenue Growth (%)"]:.1f}%',
        va='center',
        ha='left',
        fontsize=9
    )

plt.title("Average Revenue Growth by Company")
plt.xlabel("Avg Growth (%)")
plt.ylabel("Company")
plt.tight_layout()
plt.show()
```

```python
# About the results: 

# META and GOOGL lead with the highest average growth rates (~18–19%), showing aggressive expansion in revenue
# MSFT and AMZN follow, with strong, consistent growth around ~13–14%
# AAPL shows the lowest growth (~10%), but still positive and stable — likely reflecting its mature market position
```

```python
# 3. Revenue vs Net Income (Scatter Plot)

# Scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(data=latest_data, x="Revenue (B)", y="Net Income (B)", hue="Company", s=100)

# Add dashed diagonal line for reference (e.g., 15% margin)
x_vals = np.linspace(0, latest_data["Revenue (B)"].max() + 100, 100)
plt.plot(x_vals, 0.15 * x_vals, '--', color='gray', label='15% Net Margin')

# Titles & labels
plt.title("Revenue vs Net Income (2024)")
plt.xlabel("Revenue (B)")
plt.ylabel("Net Income (B)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# About the results: 

# GOOGL has the highest net income (~100B) with moderate revenue → most efficient (above 15% revenue)
# AMZN has the highest revenue (~640B) but the lowest profit (~59B)  (Well below the line)
# AAPL, MSFT, and META sit between those extremes: 
# -Apple is strong in both revenue and profit (slightly above the line)
# -Meta has lower revenue, but decent profit — suggests good margins (slightly above the line)
# Microsoft is also quite efficient with a solid net income (Above the line)
```
